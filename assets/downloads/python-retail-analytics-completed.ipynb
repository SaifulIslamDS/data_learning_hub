```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
DATA = Path('.')
```

```python
sales = pd.read_csv(DATA / 'python_retail_sales.csv', parse_dates=['order_date'])
print(sales.shape)
print(sales.isna().sum().sort_values(ascending=False).head())
print('Duplicates:', sales.duplicated(['order_id','product_id']).sum())
```

```python
summary = sales.groupby('region', as_index=False).agg(orders=('order_id','nunique'), revenue=('revenue','sum'), profit=('profit','sum'))
summary['margin'] = summary['profit']/summary['revenue']
summary.sort_values('revenue', ascending=False)
```

```python
monthly = sales.set_index('order_date').resample('MS').agg(revenue=('revenue','sum'),profit=('profit','sum'))
ax = monthly['revenue'].plot(title='Monthly revenue', figsize=(9,4))
ax.set_ylabel('BDT')
plt.show()
```
